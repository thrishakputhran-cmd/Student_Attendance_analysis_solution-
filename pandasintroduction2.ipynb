## Dealing with Duplicates values, Missing Data

```python
import pandas as pd
```

### Duplicates

```python
cdata=pd.read_csv("subdata_pandas/company.csv")
cdata
```

```python
#check Boolean format
cdata.duplicated()
```

```python
#select duplicate row
cdata[cdata.duplicated()]
```

```python
cdata["EEID"].duplicated() #Boolean
```

```python

cdata[cdata["EEID"].duplicated()]
```

```python
cdata["Name"].duplicated()
```

```python
cdata[cdata["Name"].duplicated()]
```

```python
#Drop duplicate
cdata.drop_duplicates()
```

```python
cdata["EEID"].drop_duplicates()
```
