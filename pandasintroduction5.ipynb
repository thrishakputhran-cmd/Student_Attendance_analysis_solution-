### Compare Pivoting and Melting Dataframe

```python
import pandas as pd
```

### Compare

```python
shop_stock_06Oct={
    'Fruits':['Mango','Apple','Banana','Papaya'],
    'Price':[100,150,60,30],
    'Qunt':[10,15,6,3],

}
stock_06Oct_df=pd.DataFrame(shop_stock_06Oct)
shop_stock_07Oct={
    'Fruits':['Mango','Apple','Banana','Papaya'],
        'Price':[150,200,60,35],
        'Qunt':[15,12,12,8],

}
stock_07Oct_df=pd.DataFrame(shop_stock_07Oct)
```

```python
#-------------Compare--------------
stock_06Oct_df.compare(stock_07Oct_df)
```

```python
stock_06Oct_df.compare(stock_07Oct_df,align_axis=0)
```

```python
#-------------compare-----------
#self                       #others
stock_06Oct_df.compare(stock_07Oct_df)
stock_06Oct_df.compare(stock_07Oct_df,align_axis=0)
stock_06Oct_df.compare(stock_07Oct_df,keep_equal=True)
stock_06Oct_df.compare(stock_07Oct_df,keep_shape=True)
stock_06Oct_df.compare(stock_07Oct_df,align_axis=0,keep_equal=True,keep_shape=True)
```

```python
stock_06Oct_df.compare(stock_07Oct_df,keep_equal=True)
```

```python
stock_06Oct_df.compare(stock_07Oct_df,keep_shape=True)
```

```python
stock_06Oct_df.compare(stock_07Oct_df,align_axis=0,keep_equal=True,keep_shape=True)
```

```python
# Pivoting

shop_sales={
    'Product':['laptop','mobile','laptop','mobile','laptop','mobile','laptop'],
    'Month':['jan','jan','feb','feb','mar','mar','apr'],
    'Price':[50000,12000,65000,25000,70000,35000,70000],
}
sales_df=pd.DataFrame(shop_sales)
sales_df
```

```python
#------------Pivot----------
sales_df.pivot(
    index="Product",
    columns="Month",
    values="Price"

)
```

```python
# Pivoting

shop_sales={
    'Product':['laptop','mobile','laptop','mobile','laptop','mobile','laptop'],
    'Month':['jan','jan','feb','feb','mar','mar','mar'],
    'Price':[50000,12000,65000,25000,70000,35000,70000],
}
sales_df=pd.DataFrame(shop_sales)
sales_df
```

```python
#------------Pivot----------
sales_df.pivot_table(
    index="Product",
    columns="Month",
    values="Price",
    aggfunc="sum"

)
```

### Melting

```python
sales_data={
    "product":['laptop','mobile','camera'],
    "jan":[145000,50000,265000],
    "feb":[245000,10000,365000],
    "mar":[345000,11000,465000]
}
sales_df=pd.DataFrame(sales_data)
```

```python
sales_df.melt(
    id_vars="product",
    var_name="month",
    value_name="sales"
)
```
