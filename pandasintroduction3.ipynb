Column Transformation in pandas

```python
import pandas as pd
```

Rows and col selection

```python
sdata=pd.read_csv("subdata-pandas/student.csv")
sdata
```

```python
#select column from table
sdata[['firstName','marks','fees']]
```

```python
#select rows based on condition
sdata.loc[sdata.id==106]
sdata.loc[sdata.firstName=="Pooja"]
sdata.loc[sdata.age>20]
```

```python
#slicing-rows loc[strt index:endidex]
sdata.loc[2:6]
#slicing on-rows and col iloc[strtIdx: endIdx-1,startIdx:endIdx-1]
sdata.iloc[1:5,1:4]
```

Filter

```python
sdata=pd.read_csv("subdata-pandas/student.csv")
sdata
```

```python
sdata['age']>20 #boolean
sdata[sdata['age']>20]
sdata_age_filter=sdata[sdata['age']>20]
sdata_age_filter
```

```python
sdata.where(sdata['age']>20,other="not eligible")
sdata.where(sdata['marks']>=85,other="not selected")
```

```python
sdata.loc[sdata['marks']>=85, "Result"]="selected"
sdata.loc[sdata['marks']<85, "Result"]="Not selected"
sdata
```
