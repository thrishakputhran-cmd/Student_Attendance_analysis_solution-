GroupBy Merge or Join 

```python
import pandas as pd
```

```python
exp=pd.read_excel("subdata_pandas/expense.xlsx")
exp.head(5)
```

```python
exp.groupby('Category')['Amount'].sum()#type1
```

```python
exp.groupby('Category')['Amount'].agg([sum])#type2
```

```python
exp.groupby('Category')['Amount'].agg(['sum','count','mean','max','min'])#give a complete expense summary for each category.
```

```python
# how much money was paid using each payment mode
exp.groupby("Payment Mode")['Amount'].sum()
```

```python
#how many transaction were made using each payment mode?
exp.groupby("Payment Mode")["Amount"].agg(['count'])
```

```python
#for each payment mode, calculate the  total amount spent across
exp.groupby(['Payment Mode','Category'])['Amount'].agg(['count'])
```

```python
#which categories have the highest total expenses?
exp.groupby(['Payment Mode','Category'])['Amount'].sum().sort_values(ascending=False)
```

```python
#for each category how much amount spend using each payament mode
exp.groupby(['Category','Payment Mode'])['Amount'].sum()
```

### Merge Or Join

```python
customer=pd.read_csv("subdata_pandas/customer.csv")
customer
```

```python
payment=pd.read_csv("subdata_pandas/payment.csv")
payment
```

```python
#inner join
pd.merge(left=customer,right=payment,on='customerid',how='inner')
```

```python
#left join
pd.merge(left=customer,right=payment,on='customerid',how='left')
```

```python
#Right join
pd.merge(left=customer,right=payment,on='customerid',how='right')

```

```python
exp.groupby('Category')['Amount'].sum().reset_index()
exp.groupby('Payment Mode')['Amount'].mean().reset_index()
exp.groupby('Category')['Amount'].max().reset_index()
exp.groupby('Payment Mode')['Amount'].agg(['sum', 'mean', 'max']).reset_index()
```

```python
exp.groupby('Category')['Amount'].sum().reset_index()
```

```python
exp.groupby('Payment Mode')['Amount'].mean().reset_index()
```

```python
exp.groupby('Category')['Amount'].max().reset_index()
```

```python
exp.groupby('Category')['Amount'].max().reset_index()
```

```python
# 1. Total expense of each category
exp.groupby('Category')['Amount'].sum()
```

```python
# 2. Average transaction amount for each payment mode
exp.groupby('Payment Mode')['Amount'].mean()
```

```python
# 3. Highest individual transaction in each category
exp.groupby('Category')['Amount'].max()
```
